# Strings and string formatting
A scientific data analysis workflow almost invariably implies working with strings, e.g., for filepaths (more details in the future) and when writing output to files and the command line. Fortunately, Python comes with many handy tools to format strings.

Strings can be enclosed in single quotes ('...') or double quotes ("...") with the same result:

````python
'spam eggs'
````

````python
"spam eggs"
````

The double quotes are useful if you need to use a single quote in a string:

````python
'doesn\'t'
````

Alternatively, `\` can be used to escape quotes:

````python
'doesn\'t'
````

If you do not want characters prefaced by `\` to be interpreted as special characters, you can use raw strings by adding an `r` before the first quote. This is useful for Windows paths:

````python
print('C:\some\name')  # here \n means newline!
````

````python
print(r'C:\some\name')  # note the r before the quote
````

```{admonition} For Windows users
:class: warning 

**Windows users**: remember this trick! Paths to files or folders are used constantly in programming.
```

## String formatting
"String formatting" refers to formatting the content of variables (strings, numbers, paths, etc.) into strings, for example to display them on screen or to write them to a text file. Unfortunately, there is more than one way to format strings in Python (actually there are at least 4!). This short section will guide you to the ones you should use preferably.

### The modern way: formatted string ("f-string") literals
Consider the following example:

````python
name = 'Assane'
print(f'Hello {name}!')
````

The important bit here is the **`f` prefix** to the string literal which indicates to the interpreter that the string might contain **curly braces**, which contain variable names that will be replaced with their values.

Not only strings can be formatted into a string. Numbers can too:

````python
n = 3
print(f'{name} has {n} cats.')
````

````python
pi = 3.14 
print(f'pi ≈ {pi}')
````

f-strings are quite powerful. They can evaluate arbitrary expressions:

````python
print(f'2 pi ≈ {2 * pi}')
````

This feature is to be used with care, you probably do not want to have very complicated expressions within your f-strings!

#### Formatting numbers in strings
Very often, you want your strings to be of predictable length and format. For example, you may want float numbers to be printed only with a chosen precision:

````python
frac = 2 / 3
print(f'Set free: {frac}')
print(f'Formatted to 2 decimals: {frac:.2f}')  # f is for "float"
print(f'Formatted to an integer: {int(frac)}')
print(f'Formatted to a rounded integer: {round(frac)}')
print(f'Formatted to a rounded integer with leading spaces: {round(frac):4d}')  # d is for "int"
print(f'Formatted to a rounded integer with leading zeros: {round(frac):04d}')
````

#### Formatting dates in strings 
Formatting dates is also possible:

````python
import datetime
now = datetime.datetime.now()

print(now)
print(f'{now:%Y-%m-%d %H:%M}')
````

### The old ways to format strings
f-strings have been added to Python in version 3.6 (end of 2016). Before that, two other string formatting tools were available and are partly still used today. Therefore, you should be familiar with them as well.

#### The "not too old" way: `.format()` (still useful!)
This string formatting method was introduced with Python 3 (so that is already a bit older). It works with appending a call to the `.format()` method to a string:

````python
print('Hello, {}!'.format(name))
````

No `f` prefix here (do not mix them up!). Otherwise, it works more or less the same:

````python
print('{} has {} cats.'.format(name, n))
print('{name_of_person} has {n_cats:02d} cats, right?'.format(n_cats=n, name_of_person=name))
````

You see where this is going! This is pretty much the same syntax, but a bit more verbose. f-strings are generally more readable and should be preferred, but the `.format()` can be useful in very specific cases, as shown example below. 

````python
template_string = 'Hey {name}! I think you should come over to {city} and visit {place}.'
# some code ommitted here...
print(template_string.format(name='Lesedi', city='Cape Town', place='Table Mountain'))
````

The nice thing in the example above is that you can create a template string early in your script, that you can "fill" with values later in your program.

#### The very old way: the `%` operator
This was the standard in Python 2 but is still working today (and will continue to work in the future). You may encounter it some older code.

````python
print('Hello %s!' % name)
````
