# File paths
Handling file paths is one of the maybe not-so-fun parts of a scientist's job. But we have to work with file paths all the time, e.g., to read in data files, write new data files, or save the plots we created.

Technically, you could simply use everything we learned about string formatting to manually create a file path, e.g., the file where want to save your new figure:

```python
figure_directory = '/home/username/my_figures/'
figure_temp = figure_directory + 'temp_timeseries.png'
print(figure_temp)
```

The problem will occur when you want to run your code on different platforms, e.g., Windows and Linux since Windows uses backslashes and Linux forward slashes for file paths. For this reason, you should never hardcode your file paths, but rather use one of the existing packages for handling file paths.

## pathlib

`pathlib` is the current standard for handling file paths. To concatenate the components of the path with `pathlib`, we first have to create a `Path` object:


```python
from pathlib import Path

data_directory = Path('/home/username/my_data/')
print(data_directory)
type(data_directory)
```

Now we can simply use forward slashes to concatenate the components of our file path, which is a very intuitive way. Always use forward slashes, `pathlib` will take care of the rest.

```python
file_temp = data_directory / 'data_temp.csv'
print(file_temp)
type(file_temp)
```

## The old way: os.path

The os.path module is still used a lot to handle file names. It includes the method `join()` which allows you to build platform-specific file paths.

```python
import os.path

figure_temp = os.path.join(figure_directory, 'temp_timeseries.png')
print(figure_temp)
```

In contrast to `pathlib`, `os.path.join` does not create a `Path()` object, but a string.

```python
type(figure_temp)
```

For some additional information, you can read this [short blog post](https://medium.com/@ageitgey/python-3-quick-tip-the-easy-way-to-deal-with-file-paths-on-windows-mac-and-linux-11a072b58d5f) which is a good entry-level tutorial to path handling.

### Take home: path handling
- Never use `+` and other string shenanigans to handle your file paths.
- [pathlib](https://docs.python.org/3/library/pathlib.html) is the current standard. It provides a very intuitive way of path handling once you get used to it.
- [os.path](https://docs.python.org/3/library/os.path.html) and, in particular, `os.path.join` is another simple way to deal with paths as strings.
